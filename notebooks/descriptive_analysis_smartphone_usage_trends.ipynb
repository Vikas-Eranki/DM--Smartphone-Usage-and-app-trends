***Descriptive Analysis***

***About Dataset***

***This descriptive analysis aims to understand smartphone usage behavior across different user demographics. The dataset includes metrics such as screen time, total app usage, category-wise app usage (social media, productivity, gaming), and number of apps used, along with demographic attributes like age, gender, and location.***


* ***User_ID***: *A unique numeric identifier assigned to each participant in the mobile usage dataset.*

* ***Age***: *The age of the user in years, representing the demographic distribution across young, adult, and older populations.*

* ***Gender***: *The gender identity of the user (Male/Female), used for studying gender-wise variations in mobile usage behavior.*

* ***Total_App_Usage_Hours***: *The total number of hours the user spends using different apps daily, summing all app categories.*

* ***Daily_Screen_Time_Hours***: *The total screen-on time per day, including active usage, browsing, and passive device engagement.*

* ***Number_of_Apps_Used***: *The total count of mobile applications actively used by the user within the observed period.*

* ***Social_Media_Usage_Hours***: *Daily time (in hours) spent on social media platforms such as Instagram, Facebook, Snapchat, and similar apps.*

* ***Productivity_App_Usage_Hours***: *Daily time (in hours) spent on productivity-oriented apps such as calendars, note-taking apps, emails, and task managers.*

* ***Gaming_App_Usage_Hours***: *Daily time (in hours) spent on gaming apps, ranging from casual mobile games to high-engagement gaming applications.*

* ***Location***: *The city or region where the user resides, used to analyze geographic differences in smartphone usage behavior.*

```python
def compute_stats(series):
    data = series.dropna().values

    mean = np.mean(data)
    median = np.median(data)
    mode = stats.mode(data, keepdims=True)[0][0]
    std = np.std(data)

    p25 = np.percentile(data, 25)
    p75 = np.percentile(data, 75)
    iqr = p75 - p25
    lower = p25 - 1.5 * iqr
    upper = p75 + 1.5 * iqr

    outliers = data[(data < lower) | (data > upper)]

    z = stats.zscore(data)
    anomalies = data[np.abs(z) > 3]

    return {
        "mean": mean,
        "median": median,
        "mode": mode,
        "std_dev": std,
        "iqr": iqr,
        "outliers_count": len(outliers),
        "anomalies_count": len(anomalies)
    }
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
```

```python
numerical_col_results = {}

for col in num_cols:
    numerical_col_results[col] = compute_stats(df[col])

numerical_col_results
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
```

```python
encoder = LabelEncoder()
df_cat_encoded = df[categorical_cols].apply(encoder.fit_transform)
```

```python
scaler = StandardScaler()
df_cat_scaled = pd.DataFrame(scaler.fit_transform(df_cat_encoded), columns=categorical_cols)
```

```python
categorical_results = {}

for col in df_cat_scaled.columns:
    categorical_results[col] = compute_stats(df_cat_scaled[col])

categorical_results
```
